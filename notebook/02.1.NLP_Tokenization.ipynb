```python
text=  "On wednesday, the U.S. Food and Drug Administration approved a new drug for the treatment of Alzheimer's disease. The drug, called Aduhelm, is the first new treatment for Alzheimer's in nearly two decades. It works by targeting amyloid plaques in the brain, which are believed to contribute to the progression of the disease. The approval has sparked both excitement and controversy within the medical community, as some experts question the drug's effectiveness and potential side effects."
```

```python
nltk.download('punkt')
```

```python
from nltk.tokenize import sent_tokenize
sent_tk=sent_tokenize(text)
print("Sentence tokeizint the text: \n")
print(sent_tk)
```

```python
from nltk.tokenize import word_tokenize
word_tk=word_tokenize(text)
print("\nWord tokenizing the text: \n")
print(word_tk)
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
sw=set(stopwords.words('english'))

print("Stop words in English are:\\n")
print(sw)
```

```python
filtered_words=[word for word in word_tk if not word in sw]
print("\nFiltered words after removing stop words: \n")
print(filtered_words)
```

```python
from nltk.stem import PorterStemmer
port_size = PorterStemmer()
```

```python
stemmed_words=[]
for word in filtered_words:
    stemmed_words.append(port_size.stem(word))

print("Filtered sentence: \n", filtered_words, "\n")
print("\nStemmed words: \n")
print(stemmed_words)
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem.wordnet import WordNetLemmatizer
lem=WordNetLemmatizer()

lemm_words=[]

for i in range(len(filtered_words)):
    lemm_words.append(lem.lemmatize(filtered_words[i]))

print("Lemmatized words: \n")
print(lemm_words)
```
